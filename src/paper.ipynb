\vspace*{\fill}

# Introduction

In recent years, social media platforms such as YouTube, Instagram and TikTok have significantly increased the viewership, exposure and global reach of dance. The International Dance League (IDL), the first global professional dance league, has drawn millions of views since its 2025 founding, and its 2026 season now airs on ESPN2 and Disney+ (International Dance League, 2026c).

The league consists of 6 pro dance teams competing across 6 events in a regular season. Each event has two rounds: in Round 1, the 6 teams compete in three 2v2 matches; in Round 2, the three Round 1 winners compete in a 3-way match (International Dance League, 2026a). 

A team’s score combines scores from a 6-judge panel on choreography and execution categories with a fan vote component. In Round 1 matches, each judge awards 1 point to the team they scored highest on their scoresheet and 1 fan vote point is awarded to the team with higher fan vote share. The team with the majority of the 7 possible points wins the match. In Round 2, each team's judge score is the average of the 6 judges’ total scores; a share of 1 point, proportional to their fan vote share, is added to their judge score to produce a final score. Teams are ranked 1st, 2nd, and 3rd by this final score, earning 7,5, and 3 points respectively. A season's championship standings are based on cumulative points summed across all events. Per event, a team earns points from its Round 1 match outcome plus Round 2 ranking points (International Dance League, 2026a).  

Despite contributing a small share of a team's combined score in each round, fan vote can still influence match outcomes and season standings.

In Round 1, fan vote can break a 3-3 judge tie: 2 of the 12 (17%) matches played so far were decided this way. Fan vote can affect season standings when judges are unanimous: in 1 of 12 round 1 matches so far (8%), a team lost the judge vote 0-6 but still earned a point toward the season standings by winning the fan vote. In Round 2, fan vote can only affect rankings when two teams' averaged judge scores are separated by less than 1 point. This gap occurred in 1 of the 4 (25%) round 2 matches so far, but the fan vote outcome aligned with the judge-score ranking, leaving their ranks unchanged (International Dance League, 2026d).

Fan votes are collected through the IDL web application at app.idl.pro, with voting open to both in-venue attendees and online livestream viewers. Voting opens after all teams in a match have performed and remains open for a 60 second window. Voter authentication in the app prevents duplicate votes (International Dance League, 2026a). 

In-person voters are capped by venue capacity. Total votes increasingly exceed venue capacity as shown in Results Table 1, suggesting a growing share of fan votes come from online livestream viewers.

IDL streams live across four platforms: YouTube, Twitch, TikTok, and SOOP (International Dance League, 2026b). Livestream viewers express their opinions about the matches in real time through comments,  producing time-stamped data that can be analyzed to offer teams a more detailed source of feedback than fan vote share.

This paper examines whether fan votes agree with live chat sentiment and how both compare to judges' scores, with the goal of helping teams understand how to balance choreography choices aimed at judges versus those aimed at fans.

## Relavant Prior Work

Fan votes are used in major sport leagues to decide players for all-star games. Kunz-Kaltenhäuser et al. (2023) compared voting patterns across the NFL's All-Pro awards and Pro Bowl selections  (expert voting) before and after the introduction of public balloting (public voting), using panel probit regression to model each award's selection outcome as a function of standardized performance markers (e.g., touchdowns, yards, sacks) and non-performance markers (e.g., age, previous awards, team win-loss record). In MMA, fans can submit their opinion by scoring the bout on the same 10-point scorecard used by judges, retrospectively. Berthet (2024) compared judges' standard scores and consensus scores against fan-submitted scorecards to measure how closely each scoring method aligned with fan opinion using percent agreement. To the best of our knowledge, sports literature has not yet examined judge-fan agreement within a live match whose outcome directly affects season standings.

Sentiment analysis is widely used to study sport fans. Chang (2019) used a lexicon-based sentiment classifier to classify real-time tweets posted during the Super Bowl 50 game into discrete emotion categories which aligned with live game scoring events. Positive emotion word counts increased after a fan's team scored and negative emotions word counts increased after the opposing team scored.  Kobs et al. (2020) built lexicon-based sentiment classifiers for Twitch-specific slang and emotes and used their outputs to weakly train a neural network, which they applied to plot sentiment over time for a live-streamed game convention keynote, showing a sharp sentiment drop when the audience learned that the newly announced game, Diablo Immortal, would be released for smartphones only. However, these studies interpret the sentiment signal by aligning it to live game  or stream events; whether sentiment agrees with an independent fan vote outcome has not been studied in a sports context.

No formal research on the International Dance League currently exists. IDL's format allows judges, an official fan vote, and real-time chat sentiment to be observed simultaneously for the same performance. Therefore, our paper addresses this gap by examining agreement between fan votes,  judges' scores, and live chat sentiment across IDL matches. 

# Methods

## Dataset

Event metadata, judges' scores, and fan votes of 4 events completed so far in the 2026 season were scraped from the IDL website (International Dance League, 2026d). The dataset includes 12 Round 1 matches and 4 Round 2 matches. We manually labeled each venue's maximum capacity, shown in Results Table 1, using figures reported on the respective venue websites (Manhattan Center, n.d.; Playbill Venues, 2026; Time Out, n.d.; UBC Sport Facilities, n.d.).

We restrict our sentiment analysis to YouTube because it has the largest share of IDL's livestream audience. IDL streams on two YouTube channels, @idl.global (70K followers; International Dance League, n.d.-a) and @steezy (1.97 million followers; STEEZY Studio, n.d.). IDL's Twitch channel has 1,200 followers (International Dance League, n.d.-b), TikTok channel has 170 (International Dance League, n.d.-d), and SOOP channel has 0 (International Dance League, n.d.-c). Excluding other platforms may underrepresent segments with different demographics and threaten external validity. However, the majority of IDL's livestream audience is on YouTube and it can provide valuable insight about the majority of the fans.

Live chat replay must be enabled in order to access live chat comments after a stream ends in Youtube. Since IDL did not enable this feature for most of its livestreams, our live chat data is limited to the @idl.global channel's Seoul event and the @steezy channel's Seoul and Sydney events. We scraped a total of 22,918 live chat comments from these livestreams. IDL has communicated to us that live chat replay will be enabled for all future livestreams. 

We manually labeled the start and end timestamps of each team's routine within a match using the corresponding YouTube broadcasts, then restricted the dataset to comments posted during a team's performance window. In total, 3,935 live chat comments were posted during teams' performance windows.

## Sentiment Classification


IDL YouTube live chat comments are very different from standard English. The comments consist of different languages, emojis, dance team or player shoutouts, dance-specific slang, spelling errors, and abbreviations. We use an existing sentiment classification model, AmaanP314/youtube-xlm-roberta-base-sentiment-multilingual (AmaanP314, 2024), as our baseline. The model was pre-trained on approximately 198 million multilingual tweets and fine-tuned on over 1 million YouTube comments.

## Labelled Dataset

To evaluate the model's performance on IDL YouTube live chat comments, we created a sample of 500 live chat comments manually labeled with sentiment.

To create a representative sample, we removed duplicate comments and retained 3,267 unique comments. Comments varied across teams in content and volume. The sample was drawn using stratified sampling by team to prevent over-representation from teams with higher comment volume.

Comments were manually labeled as positive, negative, or neutral based on the sentiment expressed during each team's performence. Labeling was performed by two labelers with a dance background, chosen for their familiarity with dance terminology, abbreviations, and community slang. Each labeler annotated a distinct set of 250 comments. Label consistency across annotators was not formally quantified.

Sentiment was assigned using a set of labeling rules:
- Each comment was judged individually instead of being matched against fixed keyword patterns
- Non-English comments were translated with Google Translate and labeled based on the translated meaning
- Comments referencing non-dance elements such as camerawork, costumes, staging, and music choice were labeled according to the sentiment expressed
- Comments were labeled neutral when they were unrelated to sentiment (e.g., "where do I vote?"), when they mentioned a team name without accompanying sentiment language, and when a translation was unclear

Limitations of this labeling process include inter-labeler subjectivity in sentiment judgments, particularly for mixed or ambiguous comments, and potential loss of nuance or slang meaning in translated comments. Additionally, the labels reflect the interpretations of only two labelers, they may not generalize to broader audience consensus.

## Baseline model evaluation

We evaluated the baseline model, AmaanP314/youtube-xlm-roberta-base-sentiment-multilingual, on our labeled sample. The model achieved an overall accuracy of 0.479 and a macro F1 score of 0.470, with macro recall of 0.513.

The model had low recall for the positive class (0.412) and missed a large share of comments that were actually positive. It also had low precision for the neutral class (0.297) and frequently predicted "neutral" even when the true label was positive or negative. Negative comments were classified fairly accurately, and both precision (0.449) and recall (0.460) were reasonably balanced for that class.

The baseline model struggled to distinguish neutral sentiment from positive or negative sentiment.

## Fine tuning

We fine-tuned AmaanP314/youtube-xlm-roberta-base-sentiment-multilingual on our labeled sample to improve performance on IDL YouTube live chat data.

The labeled sample has a class imbalance, with positive comments dominating (59%), followed by neutral (24%) and negative (17%). Given the small dataset size and this class imbalance, we used stratified 5-fold cross-validation rather than a single train/test split, allowing every labeled comment to be used for both training and evaluation while preserving class distribution across folds. 

Standard classifiers trained on imbalanced data tend to be biased toward the majority class (He & Garcia, 2009), so we used a class-weighted cross-entropy loss during training. Each class was assigned a weight equal to the inverse of its frequency in the training fold, so that errors on less frequent classes contributed more heavily to the overall loss than errors on more common classes (Huang et al., 2016). 

## Fine tuned model evaluation

After fine-tuning, the model achieved a mean macro F1 of 0.649 (± 0.024) across folds, accuracy of 0.705 and macro recall of 0.652. All three metrics improved substantially from the baseline (macro F1: 0.470, accuracy: 0.479, macro recall: 0.513), indicating that fine tuning on IDL-specific live chat data meaningfully improved the model's ability to distinguish sentiment classes.

We then used the fine-tuned model to label sentiment across the full set of live chat comments posted during teams' performance windows.

# Results

![](../results/tables_and_figures/table1_event_summary.png)

To assess whether fan vote share and live chat sentiment agree, we used Kendall's tau to measure rank agreement between fan vote share and net sentiment from live chat comments. We computed Kendall's tau separately for Round 1 and Round 2, since Round 1 matches are 2-way and Round 2 matches are 3-way. An even split corresponds to 50% per team in a 2-way match and 33% per team in a 3-way match; pooling the two rounds would combine shares computed against different reference points.

Net sentiment and fan vote share are not directly comparable in raw form. Fan vote share is a bounded proportion that sums to 1 within a match, while net sentiment is an unbounded score that can be negative. We applied a softmax transformation to net sentiment within each match, converting it into a sentiment share that also sums to 1 across teams. 

Across the Seoul and Sydney events, fan vote share and sentiment share showed a weak positive association (tau = 0.303, 95% CI [-0.328, 0.747]) in Round 1, while Round 2 showed a weak negative association (tau = -0.105, 95% CI [-0.904, 0.857]). Neither association was statistical significant and the wide confidence intervals reflect the limited sample size.

To assess whether fan vote share and judges' scores agree, we repeat the approach of computing Kendall's tau between fan vote share and judges share separately for Round 1 and Round 2. Judges match average scores are also not directly comparable to fan vote share in raw form, so we applied the same softmax transformation to judges match average scores within each match, converting them into a judges share that sums to 1 across teams.

Fan vote share and judges share showed a strong positive association in Round 1 (tau = 0.923, 95% CI [0.743, 0.979]), which was statistically significant. Round 2 also showed a positive association (tau = 0.600, 95% CI [-0.412, 0.949]), though it did not reach statistical significance, reflecting the smaller sample size.

![](../results/tables_and_figures/fig1_category_tau_heatmap.png){height=3in}

We examined fan vote share and judges' scores agreement at the judge scores category level. Agreement was positive across nearly every category in both rounds, consistent with the strong overall fan vote-judges association reported above. Agreement was strongest for Creativity (tau = 0.65 in Round 1, 1.00 in Round 2), Projection/Communication (tau = 0.76 and 0.83), and Staging (tau = 0.49 and 0.83), suggesting fans and judges are especially aligned on these dimensions.

We repeated this approach once more to assess whether judges' scores and live chat sentiment agree, computing Kendall's tau between judges share and sentiment share separately for Round 1 and Round 2.

Judges share and sentiment share showed a weak positive association in both Round 1 (tau = 0.215, 95% CI [-0.409, 0.702]) and Round 2 (tau = 0.105, 95% CI [-0.857, 0.904]). Neither association reached statistical significance, and the wide confidence intervals reflect the limited sample size.

![](../results/tables_and_figures/fig2_category_tau_heatmap_sentiment.png){height=3in}

Sentiment agreement with individual judge categories was weaker and less consistent than fan vote agreement with those same categories. The strongest association was Spacing in Round 1 (tau = 0.71), followed by Musicality, which showed a moderate negative association (tau = -0.48), the only category where sentiment moved opposite to the judged score. Most other categories showed weak associations in either direction. 

Several categories, including Choreography Complexity, Staging, and Technical Execution + Authenticity, showed opposite association directions between Round 1 and Round 2. This inconsistency suggests live chat sentiment tracks specific judged categories far less reliably than the fan vote does.

# Conclusion

Across the Seoul and Sydney events, judges and fan votes showed strong agreement in Round 1 (tau = 0.923) that weakened in Round 2 (tau = 0.600), while live chat sentiment showed only weak, inconsistent agreement with both fan votes and judges' scores in both rounds. At the category level, fan vote agreement with judges was positive and comparatively strong across nearly every judged category, while sentiment agreement with judges was weaker and less consistent, even reversing direction between rounds for several categories. Overall, judges and fans appear to evaluate routines in a broadly similar way, while live chat sentiment does not yet serve as a reliable proxy for either.

### Future Work

These results are based on a small sample, so all associations should be treated as preliminary. As more IDL events are completed, this analysis can be repeated with greater statistical power to determine whether the patterns observed here hold and to support comparisons between judges, fan vote, and sentiment agreement across specific scoring categories, in order to optimize which aspects of a routine most resonate with judges versus fans versus the live audience.
